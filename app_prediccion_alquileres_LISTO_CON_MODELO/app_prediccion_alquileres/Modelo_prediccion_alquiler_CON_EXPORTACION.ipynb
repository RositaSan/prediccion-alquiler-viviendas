### 4.2 Desarrollo e Implementación del Modelo

Para la implementación de los modelos se seleccionaron características estructurales y geográficas de las viviendas. La variable objetivo corresponde al precio aproximado de alquiler expresado en dólares estadounidenses (price_aprox_usd).
Las variables numéricas consideradas son la superficie total, la superficie cubierta, la latitud, la longitud, la cantidad de baños y la cantidad de ambientes. Asimismo, se incorporan variables categóricas relacionadas con el tipo de propiedad, la provincia o estado y la localidad.
Las variables categóricas serán transformadas mediante codificación One-Hot Encoding, mientras que los valores faltantes de las variables numéricas serán tratados mediante imputación utilizando la mediana.
Se compararán tres algoritmos de regresión: Regresión Lineal, Random Forest Regressor y Gradient Boosting Regressor.

```python
# 4.2 SELECCIÓN DE VARIABLES PARA EL MODELO

target = "price_aprox_usd"

numeric_features = [
    "surface_total_in_m2",
    "surface_covered_in_m2",
    "lat",
    "lon",
    "Num_Banos",
    "Num_Ambientes"
]

categorical_features = [
    "Tipo_Propiedad_Limpia",
    "state_name",
    "place_name"
]

features = numeric_features + categorical_features

# Separar variables predictoras y variable objetivo
X_train = df_train[features].copy()
y_train = df_train[target].copy()

X_test = df_test[features].copy()
y_test = df_test[target].copy()

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

print("\nVariable objetivo:", target)
print("\nVariables utilizadas:")
print(features)
```

```python
print("Valores faltantes después de depurar:\n")
print(X_train.isnull().sum().sort_values(ascending=False))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Tratamiento de variables numéricas
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

# Tratamiento de variables categóricas
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                min_frequency=5
            )
        )
    ]
)

# Combinar preprocesamiento numérico y categórico
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocesamiento configurado correctamente.")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

# REGRESIÓN LINEAL
modelo_lineal = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

# RANDOM FOREST
modelo_rf = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=100,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

# GRADIENT BOOSTING
modelo_gb = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=100,
                random_state=42
            )
        )
    ]
)

print("Modelos configurados correctamente:")
print("1. Regresión Lineal")
print("2. Random Forest Regressor")
print("3. Gradient Boosting Regressor")
```

## VI. EVALUACIÓN

### 6.1 Métricas y Evaluación del Modelo

Para evaluar el desempeño de los modelos entrenados se utilizarán las métricas MAE, MSE, RMSE y R².

El Error Absoluto Medio (MAE) representa la magnitud promedio del error entre los precios reales y los precios estimados.

El Error Cuadrático Medio (MSE) penaliza con mayor intensidad aquellas predicciones que presentan errores elevados.

La Raíz del Error Cuadrático Medio (RMSE) expresa el error en las mismas unidades de la variable objetivo, en este caso dólares estadounidenses.

Finalmente, el coeficiente de determinación R² permite analizar qué proporción de la variabilidad de los precios puede ser explicada por cada modelo.

Para MAE, MSE y RMSE, los valores más bajos representan un mejor desempeño, mientras que para R² los valores más cercanos a 1 indican un mejor ajuste predictivo.


```python
# 6.1 EVALUACIÓN DE LOS MODELOS

import numpy as np
import pandas as pd

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

def evaluar_modelo(nombre, y_real, y_pred):
    mae = mean_absolute_error(y_real, y_pred)
    mse = mean_squared_error(y_real, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_real, y_pred)

    return {
        "Modelo": nombre,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    }


resultados = []

resultados.append(
    evaluar_modelo(
        "Regresión Lineal",
        y_test,
        pred_lineal
    )
)

resultados.append(
    evaluar_modelo(
        "Random Forest",
        y_test,
        pred_rf
    )
)

resultados.append(
    evaluar_modelo(
        "Gradient Boosting",
        y_test,
        pred_gb
    )
)

df_resultados = pd.DataFrame(resultados)

df_resultados
```

```python
# Ordenar modelos desde el menor error al mayor

df_resultados_ordenados = df_resultados.sort_values(
    by="MAE"
).reset_index(drop=True)

print("Comparación de modelos ordenados por MAE:")
display(df_resultados_ordenados)
```

```python
# COMPARACIÓN GRÁFICA DEL MAE

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    df_resultados["Modelo"],
    df_resultados["MAE"]
)

plt.xlabel("Modelo")
plt.ylabel("MAE (USD)")
plt.title("Comparación del Error Absoluto Medio")

plt.xticks(rotation=15)

plt.show()
```

### Validación cruzada de los modelos

Con el propósito de obtener una evaluación más robusta y reducir la dependencia de una única partición de los datos, se aplicó validación cruzada de 5 folds sobre el conjunto de entrenamiento.

En cada iteración, cuatro particiones son utilizadas para entrenar el modelo y una para validarlo. El proceso se repite hasta que cada partición haya sido utilizada como conjunto de validación.

Para comparar los modelos se calcularán el MAE, RMSE y R² promedio obtenidos durante las cinco iteraciones.


```python
from sklearn.model_selection import KFold, cross_validate
import pandas as pd
import numpy as np

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2"
}

modelos = {
    "Regresión Lineal": modelo_lineal,
    "Random Forest": modelo_rf,
    "Gradient Boosting": modelo_gb
}

resultados_cv = []

for nombre, modelo in modelos.items():

    print(f"Evaluando: {nombre}...")

    scores = cross_validate(
        modelo,
        X_train,
        y_train,
        cv=kf,
        scoring=scoring,
        n_jobs=-1
    )

    resultados_cv.append({
        "Modelo": nombre,
        "MAE_promedio": -scores["test_MAE"].mean(),
        "MAE_std": scores["test_MAE"].std(),
        "RMSE_promedio": -scores["test_RMSE"].mean(),
        "R2_promedio": scores["test_R2"].mean()
    })

df_cv = pd.DataFrame(resultados_cv)

display(df_cv)
```

```python
df_cv_ordenado = df_cv.sort_values(
    by="MAE_promedio"
).reset_index(drop=True)

print("Resultados de Cross-Validation ordenados por MAE:")
display(df_cv_ordenado)
```

### 6.2 Evaluación del desempeño del sistema

Los resultados obtenidos mediante validación cruzada de 5 folds muestran diferencias entre los modelos evaluados.

Random Forest presentó el menor MAE promedio, con aproximadamente 925.26 USD, lo que indica que obtuvo el menor error absoluto medio entre los modelos analizados.

Por otro lado, Gradient Boosting alcanzó el menor RMSE promedio, con aproximadamente 3591.81 USD, y el mayor coeficiente de determinación R², con un valor aproximado de 0.2712. Estos resultados indican que este modelo presentó un comportamiento más equilibrado frente a errores de mayor magnitud y una mayor capacidad para representar la variabilidad de los precios de alquiler.

La Regresión Lineal presentó el desempeño menos favorable durante la validación cruzada, obteniendo el mayor MAE y RMSE, además del menor valor de R².

Considerando conjuntamente las métricas obtenidas, se selecciona Gradient Boosting como el modelo de mejor desempeño general para el proyecto. Sin embargo, se reconoce que Random Forest presentó el menor error absoluto medio, por lo que también mostró un desempeño competitivo.

Los resultados evidencian que todavía existe una proporción importante de variabilidad en los precios que no es explicada por las variables utilizadas, por lo que el modelo debe interpretarse como una herramienta de estimación y no como un mecanismo para determinar un precio exacto de alquiler.
